# Notebook for visualizing parallelized method for more than two partitions
For the time being only as a function of major cycles until I get proper timing results, the full dataset settings can be found in their config files

```python
import matplotlib.pyplot as plt
import matplotlib
import numpy
import helpers
import os
import logging, sys
import math
matplotlib.rcParams['figure.figsize'] = [10, 5]

cmap = "turbo"

breakdowns = [
    "read",
    "grid",
    "degrid",
    "deconvolve",
    "other"
]

breakdowns_parallel = [
    "read",
    "grid",
    "degrid",
    "deconvolve",
    "wait",
    "other"
]

num_mc_l1 = 5
num_mc_msc = 10

def total_timings(timings, breakdowns):
    total = []
    
    for i, breakdown in enumerate(breakdowns):
        curr_timing = 0

        for timing in timings:
            curr_timing += timing[i]

        total.append(curr_timing)
        
    return total

pcols = ["forestgreen", "teal", "mediumblue", "blueviolet", "fuchsia"]
```

## Sgr A

$\ell$ = [3, 10, 35, 70]

$\delta$ = 1

```python
sgra_res_path = "../results/multipartition_results/SGRA/"

breakdown_data = []

partitions=["<4", "2-11", "9-36", "34-71", ">69"]
num_partitions = len(partitions)

for i in range(num_partitions):
    breakdown_data.append(helpers.read_csv(sgra_res_path + "multi_partition_sgra_pl1/mc_timings_breakdown_" + str(i), separate_rows=True))

for i in range(num_partitions):
    breakdown_data.append(helpers.read_csv(sgra_res_path + "multi_partition_sgra_pmsc/mc_timings_breakdown_" + str(i), separate_rows=True))
    
serial_breakdown_data = helpers.read_csv(sgra_res_path + "multi_partition_sgra_seriall1/mc_timings_breakdown", separate_rows=True)
msc_breakdown_data = helpers.read_csv(sgra_res_path + "multi_partition_sgra_serialmsc/mc_timings_breakdown", separate_rows=True)

dirty = helpers.readFits(sgra_res_path + "multi_partition_sgra_seriall1/residual_0.fits")
helpers.plotNImages([dirty], [""], cmap="turbo", hide_ticks=True)
```

### L1

```python
serial_times_sep = helpers.read_csv(sgra_res_path + "multi_partition_sgra_seriall1/mc_timings")
serial_times = []
serial_snrs = []

parallel_times_sep = helpers.read_csv(sgra_res_path + "multi_partition_sgra_pl1/mc_timings")
parallel_times = []
parallel_snrs = []
combined_snrs = []

best_parallel_recon = None
best_serial_recon = None
best_parallel_snr = -9999
best_serial_snr = -9999

gt = helpers.readFits("../data7/SGRA_full_gt.fits")

#For serial
curr_deconv = None
time_acc = float(serial_times_sep[0])
serial_times.append(time_acc)
serial_snrs.append(0)

for i in range(0, num_mc_l1):
    curr_deconv_img = helpers.readFits(sgra_res_path + "multi_partition_sgra_seriall1/deconvolved_" + str(i) + ".fits")
    if curr_deconv is None:
        curr_deconv = curr_deconv_img
    else:
        curr_deconv += curr_deconv_img
        
    time_acc += float(serial_times_sep[i+1])
    
    serial_snrs.append(helpers.compute_snr(curr_deconv, gt))
    serial_times.append(time_acc)
    
    if best_serial_snr < serial_snrs[-1]:
        best_serial_recon = curr_deconv.copy()
        best_serial_snr = serial_snrs[-1]
    
    
#parallel
curr_deconvs = [None for i in range(num_partitions)]
time_acc = float(parallel_times_sep[0])
parallel_times.append(time_acc)
for i in range(num_partitions):
    parallel_snrs.append([0])

combined_snrs.append(0)

for i in range(0, num_mc_l1):
    for j in range(num_partitions):
        curr_deconv = helpers.readFits(sgra_res_path + "multi_partition_sgra_pl1/deconv_" + str(j) + "_" + str(i) + ".fits")
        if curr_deconvs[j] is None:
            curr_deconvs[j] = curr_deconv
        else:
            curr_deconvs[j] += curr_deconv
            
        parallel_snrs[j].append(helpers.compute_snr(curr_deconvs[j], gt))

    time_acc += float(parallel_times_sep[i+1])

    curr_deconv_combined = helpers.readFits(sgra_res_path + "multi_partition_sgra_pl1/recon_combined_" + str(i) + ".fits")
    combined_snrs.append(helpers.compute_snr(curr_deconv_combined, gt))
    
    parallel_times.append(time_acc)
    
    if best_parallel_snr < combined_snrs[-1]:
        best_parallel_recon = curr_deconv_combined.copy()
        best_parallel_snr = combined_snrs[-1]
    
for i, p_snrs in enumerate(parallel_snrs):
    plt.plot(parallel_times, parallel_snrs[i], c=pcols[i], lw=3, alpha=0.1, marker='o', markersize=5)
    
plt.plot(serial_times, serial_snrs, c='crimson', label="$L1$", lw=3, marker='o', markersize=10)
plt.plot(parallel_times, combined_snrs, c='coral', label="p-$L1$", lw=3, marker='o', markersize=10)



plt.xlabel("Time (s)", fontsize=16)
plt.ylabel("S/N (dB)", fontsize=16)
#plt.yscale('log')
plt.legend(loc="lower right", fontsize=16)
plt.locator_params(nbins=5)
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)

#plt.savefig("../interleaved_parallelization_prototype/figures/sgra_timeacc.pdf", pad_inches=0.0, bbox_inches='tight')

plt.show()

helpers.plotNImages([best_serial_recon, best_parallel_recon], ["Serial", "Parallel"], same_scale=True, cmap=cmap)
helpers.plotNImages([numpy.abs(gt - best_serial_recon), numpy.abs(gt - best_parallel_recon)], \
                    ["Serial abs err", "Parallel abs err"], same_scale=True, cmap=cmap)
```
