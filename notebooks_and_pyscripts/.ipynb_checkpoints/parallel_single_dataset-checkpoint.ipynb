# This notebook shows results for the interleaved parallelization method freezing as many parameters as possible. The only varied aspects to obtain different sized datasets are:
1. Integration time (960s, 60s, 60s)
2. Number of channels (1, 1, 7)

```python
import matplotlib.pyplot as plt
import matplotlib
import numpy
import helpers
import os
import logging, sys
import math
matplotlib.rcParams['figure.figsize'] = [6, 5]

breakdowns = [
    "read",
    "invert",
    "predict",
    "deconvolve",
    "other"
]

breakdowns_parallel = [
    "read",
    "invert",
    "predict",
    "deconvolve",
    "wait",
    "other"
]

num_mc = 5

def total_timings(timings, breakdowns):
    total = []
    
    for i, breakdown in enumerate(breakdowns):
        curr_timing = 0

        for timing in timings:
            curr_timing += timing[i]

        total.append(curr_timing)
        
    return total

res_path = "../results/ri_parallelization_results/single_dataset/"
```

## Small dataset, 1969920 total, 1038045 $\mathcal{L}$, 951499 $\mathcal{H}$

```python
vl_parallel_breakdown_data = helpers.read_csv(res_path + "small_parallel/mc_timings_breakdown_0", separate_rows=True)
vh_parallel_breakdown_data = helpers.read_csv(res_path + "small_parallel/mc_timings_breakdown_1", separate_rows=True)
serial_breakdown_data = helpers.read_csv(res_path + "small_serial/mc_timings_breakdown", separate_rows=True)
msc_breakdown_data = helpers.read_csv(res_path + "small_msclean/mc_timings_breakdown", separate_rows=True)
vl_mscp_breakdown_data = helpers.read_csv(res_path + "small_msclean_p/mc_timings_breakdown_0", separate_rows=True)
vh_mscp_breakdown_data = helpers.read_csv(res_path + "small_msclean_p/mc_timings_breakdown_1", separate_rows=True)

dirty = helpers.readFits(res_path + "small_serial/residual_0.fits")
helpers.plotNImages([dirty], [""], cmap="turbo", hide_ticks=True)

compute_jacknifed_residual = False
compute_residuals = False
computed_weighted = False
```

```python
serial_times_sep = helpers.read_csv(res_path + "small_serial/mc_timings")
serial_times = []
serial_psnrs = []

msc_times_sep = helpers.read_csv(res_path + "small_msclean/mc_timings")
msc_times = []
msc_psnrs = []

parallel_times_sep = helpers.read_csv(small_res_path + "small_parallel/mc_timings")
parallel_times = []
vl_psnrs = []
vh_psnrs = []
combined_psnrs = []
combined_weighted_psnrs = []

parallel_msc_times_sep = helpers.read_csv(small_res_path + "small_msclean_p/mc_timings")
vlmsc_psnrs = []
vhmsc_psnrs = []
combinedmsc_psnrs = []
combinedmsc_weighted_psnrs = []
parallel_msc_times = []


gt = helpers.readFits("../data/SGRA_full_gt.fits")

#For serial
curr_deconv = None
time_acc = float(serial_times_sep[0])
serial_times.append(time_acc)
serial_psnrs.append(0)
for i in range(0, num_mc):
    curr_deconv_img = helpers.readFits(res_path + "small_serial/deconvolved_" + str(i) + ".fits")
    if curr_deconv is None:
        curr_deconv = curr_deconv_img
    else:
        curr_deconv += curr_deconv_img
        
    time_acc += float(serial_times_sep[i+1])

    serial_psnrs.append(helpers.compute_snr(curr_deconv, gt))
    serial_times.append(time_acc)
    
#For msclean
curr_deconv = None
time_acc = float(msc_times_sep[0])
msc_psnrs.append(0)
msc_times.append(time_acc)
for i in range(0, num_mc):
    curr_deconv_img = helpers.readFits(res_path + "small_msclean/deconvolved_" + str(i) + ".fits")
    if curr_deconv is None:
        curr_deconv = curr_deconv_img
    else:
        curr_deconv += curr_deconv_img
        
    time_acc += float(msc_times_sep[i+1])

    msc_psnrs.append(helpers.compute_snr(curr_deconv, gt))
    msc_times.append(time_acc)

if compute_residuals:
    weight_grid, model = helpers.compute_weights_griddata_by_channel("../data7/SGRA_small.ms", \
                                512, 0.00004163061962578778, 0, 0, [0])
    
if compute_jacknifed_residual:
    if weight_grid is None:
        weight_grid, model = helpers.compute_weights_griddata_by_channel("../data7/SGRA_small.ms", \
                                512, 0.00004163061962578778, 0, 0, [0])
        
    jk_resid = helpers.compute_jackknifed_residual_bychannel(model, "../data7/SGRA_small.ms",\
                                             512, 0.00004163061962578778, "uniform", 0, weight_grid, 0, 0, [0])
    helpers.write_nparr_to_fits(jk_resid.pixels.data, res_path + "small_parallel/jacknifed_resid.fits")

#parallel
curr_deconv_vl = None
curr_deconv_vh = None
time_acc = float(parallel_times_sep[0])
parallel_times.append(time_acc)
vl_psnrs.append(0)
vh_psnrs.append(0)
combined_psnrs.append(0)
combined_weighted_psnrs.append(0)

for i in range(0, num_mc):
    curr_deconv_vl_img = helpers.readFits(res_path + "small_parallel/deconv_vl_" + str(i) + ".fits")
    curr_deconv_vh_img = helpers.readFits(res_path + "small_parallel/deconv_vh_" + str(i) + ".fits")
    if curr_deconv_vl is None:
        curr_deconv_vl = curr_deconv_vl_img
        curr_deconv_vh = curr_deconv_vh_img
    else:
        curr_deconv_vl += curr_deconv_vl_img
        curr_deconv_vh += curr_deconv_vh_img
        
    time_acc += float(parallel_times_sep[i+1])
    
    if computed_weighted:
        #weight_vl = 1 / numpy.linalg.norm(helpers.compute_windowed_var(curr_deconv_vl, 5))
        #weight_vh = 1 / numpy.linalg.norm(helpers.compute_windowed_var(curr_deconv_vh, 5))
        weight_vl = 1 / helpers.compute_windowed_var(curr_deconv_vl, 5)
        weight_vh = 1 / helpers.compute_windowed_var(curr_deconv_vh, 5)

        weight_total = weight_vl + weight_vh
        weight_vl /= weight_total
        weight_vh /= weight_total

        curr_deconv_weighted = weight_vl * curr_deconv_vl + weight_vh * curr_deconv_vh
        helpers.write_nparr_to_fits(curr_deconv_weighted, res_path + "small_parallel/deconv_weighted_combined_" + str(i) + ".fits")
        
    curr_deconv_weighted = helpers.readFits(res_path + "small_parallel/deconv_weighted_combined_" + str(i) + ".fits")
    curr_deconv_combined = helpers.readFits(res_path + "small_parallel/deconv_combined_" + str(i) + ".fits")
    
    if compute_residuals:
        model.pixels.data[0, 0, :, :] = curr_deconv_vl
        vl_residual = helpers.compute_residual_bychannel(model, "../data7/SGRA_small.ms",\
                                                 512, 0.00004163061962578778, "uniform", 0, weight_grid, \
                                                 0, 0, [0])
        model.pixels.data[0, 0, :, :] = curr_deconv_vh
        vh_residual = helpers.compute_residual_bychannel(model, "../data7/SGRA_small.ms",\
                                                 512, 0.00004163061962578778, "uniform", 0, weight_grid, \
                                                 0, 0, [0])
        model.pixels.data[0, 0, :, :] = curr_deconv_combined
        combined_residual = helpers.compute_residual_bychannel(model, "../data7/SGRA_small.ms",\
                                                 512, 0.00004163061962578778, "uniform", 0, weight_grid, \
                                                 0, 0, [0])
        
        helpers.write_nparr_to_fits(vl_residual.pixels.data, res_path + "small_parallel/vl_fullresid_" + str(i) + ".fits")
        helpers.write_nparr_to_fits(vh_residual.pixels.data, res_path + "small_parallel/vh_fullresid_" + str(i) + ".fits")
        helpers.write_nparr_to_fits(combined_residual.pixels.data, res_path + "small_parallel/combined_fullresid_" + str(i) + ".fits")
        
    vl_psnrs.append(helpers.compute_snr(curr_deconv_vl, gt))
    vh_psnrs.append(helpers.compute_snr(curr_deconv_vh, gt))
    combined_psnrs.append(helpers.compute_snr(curr_deconv_combined, gt))
    combined_weighted_psnrs.append(helpers.compute_snr(curr_deconv_weighted, gt))
    parallel_times.append(time_acc)

#parallel msclean
curr_deconv_vl = None
curr_deconv_vh = None
time_acc = float(parallel_times_sep[0])

vlmsc_psnrs.append(0)
vhmsc_psnrs.append(0)
combinedmsc_psnrs.append(0)
combinedmsc_weighted_psnrs.append(0)
parallel_msc_times.append(time_acc)

for i in range(0, num_mc):
    curr_deconv_vl_img = helpers.readFits(res_path + "small_msclean_p/deconv_vl_" + str(i) + ".fits")
    curr_deconv_vh_img = helpers.readFits(res_path + "small_msclean_p/deconv_vh_" + str(i) + ".fits")
    if curr_deconv_vl is None:
        curr_deconv_vl = curr_deconv_vl_img
        curr_deconv_vh = curr_deconv_vh_img
    else:
        curr_deconv_vl += curr_deconv_vl_img
        curr_deconv_vh += curr_deconv_vh_img
        float(parallel_msc_times_sep[i+1])
    time_acc += float(parallel_msc_times_sep[i+1]) - float(parallel_times_sep[0]) if i == 0 else float(parallel_msc_times_sep[i+1])
    
    if computed_weighted:
        weight_vl = 1 / numpy.linalg.norm(helpers.compute_windowed_var(curr_deconv_vl, 5))
        weight_vh = 1 / numpy.linalg.norm(helpers.compute_windowed_var(curr_deconv_vh, 5))

        print(str(weight_vl) + " " + str(weight_vh))

        weight_total = weight_vl + weight_vh
        weight_vl /= weight_total
        weight_vh /= weight_total

        curr_deconv_weighted = weight_vl * curr_deconv_vl + weight_vh * curr_deconv_vh
        helpers.write_nparr_to_fits(curr_deconv_weighted, res_path + "small_msclean_p/deconv_weighted_combined_" + str(i) + ".fits")
        
    curr_deconv_weighted = helpers.readFits(res_path + "small_msclean_p/deconv_weighted_combined_" + str(i) + ".fits")
    curr_deconv_combined = helpers.readFits(res_path + "small_msclean_p/deconv_combined_" + str(i) + ".fits")
      
    vlmsc_psnrs.append(helpers.compute_snr(curr_deconv_vl, gt))
    vhmsc_psnrs.append(helpers.compute_snr(curr_deconv_vh, gt))
    combinedmsc_psnrs.append(helpers.compute_snr(curr_deconv_combined, gt))
    combinedmsc_weighted_psnrs.append(helpers.compute_snr(curr_deconv_weighted, gt))
    parallel_msc_times.append(time_acc)
    
plt.plot(serial_times, serial_psnrs, c='crimson', label="$L^1$", lw=3, marker='o', markersize=10)
plt.plot(msc_times, msc_psnrs, c='darkblue', label="msc", lw=3, marker='o', markersize=10)
#plt.plot(parallel_times, vl_psnrs, c='teal', label="${V}_\mathcal{L}$ recon", lw=2, marker='o')
#plt.plot(parallel_times, vh_psnrs, c='purple', label="${V}_\mathcal{H}$ recon", lw=2, marker='o')
plt.plot(parallel_times, combined_psnrs, c='coral', label="p-$L^1$", lw=3, marker='o', markersize=10)
#plt.plot(parallel_times, combined_weighted_psnrs, c='darkorange', label="combined weighted parallel recon", lw=2, marker='o')

#plt.plot(parallel_times, vlmsc_psnrs, c='powderblue', label="msclean ${V}_\mathcal{L}$ recon", lw=2, marker='o')
#plt.plot(parallel_times, vhmsc_psnrs, c='plum', label="msclean ${V}_\mathcal{H}$ recon", lw=2, marker='o')
plt.plot(parallel_msc_times, combinedmsc_psnrs, c='dodgerblue', label="p-msc", lw=3, marker='o', markersize=10)
#plt.plot(parallel_times, combinedmsc_weighted_psnrs, c='moccasin', label="msclean combined weighted parallel recon", lw=2, marker='o')

plt.xlabel("Time (s)", fontsize=16)
plt.ylabel("S/N (dB)", fontsize=16)
#plt.yscale('log')
plt.legend(loc="lower right", fontsize=16)
plt.locator_params(nbins=5)
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)

plt.show()
```
